```python
scaler = MinMaxScaler() #normalizes numerical columns
num_cols = ['Price', 'Distance', 'Landsize', 'BuildingArea']# Select numeric columns to scale
dataset[num_cols] = scaler.fit_transform(dataset[num_cols]) # Apply Min-Max Scaling
```

```python
dataset[['Price', 'Distance', 'Landsize', 'BuildingArea']].head() # View scaled values
```

```python
# Convert categorical columns into numeric using one-hot encoding
dataset = pd.get_dummies(
    dataset,
    columns=['Type', 'Method', 'Regionname'],
    drop_first=True
)
```
