```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
ds=pd.read_csv("nba.csv")
```

```python
ds.isnull().sum()
```

```python
ds["Age"].fillna(ds['Age'].mean(),inplace=True)
```

```python
ds['Position'] = ds['Position'].astype('string') 
```

```python
ds['Age']=ds['Age'].astype('int')
```

```python
ds['College'].fillna('Unknown',inplace=True)
```

```python
ds["Salary"].fillna(ds['Salary'].mean(),inplace=True)
```

```python
ds.isnull().sum()
```

```python
ds.dropna(subset=['Name','Team','Number','Position','Height','Weight'],inplace=True)
```

```python
ds.isnull().sum()
```

```python
ds['Name'] = ds['Name'].astype('string') 
```

```python
ds['Team'] = ds['Team'].astype('string') 
```

```python
ds['Position'] = ds['Position'].astype('string') 
```

```python
ds['College'] = ds['College'].astype('string') 
```

```python
height_groups = ds.groupby(ds["Height"]) #Group all who have the same height together.
height_groups.head() #From each height group, show me the first few rows
```

```python
height_groups["Salary"].describe() #For each height, give me a summary of salaries
```

```python
bins = [19,25,31,36,40] #create a list of numbers
labels = ["19-24", "25-30", "31-35", "36-40"] #names given to each range
ds["AgeGroup"] = pd.cut(ds["Age"], bins=bins, labels=labels)# Create an AgeGroup column by placing each player's age into labeled age ranges
```

```python
ds["AgeGroup"].value_counts() #counts how many players falls into each groups
```

```python
age_groups = ds.groupby("AgeGroup") #group the dataset by AgeGroup
```

```python
age_groups["Salary"].describe() #shows the summary of salary
```
