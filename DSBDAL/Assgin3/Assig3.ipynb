```python
import pandas as pd
import numpy as np
```

```python
ds=pd.read_csv("nba.csv")
```

```python
ds.info()
```

```python
ds.head()
```

```python
ds.isnull().sum()
```

```python
ds["Age"].fillna(ds['Age'].mean(),inplace=True)
```

```python
ds['Position'] = ds['Position'].astype('string') 
```

```python
ds['Age']=ds['Age'].astype('int')
```

```python
ds['College'].fillna('Unknown',inplace=True)
```

```python
ds["Salary"].fillna(ds['Salary'].mean(),inplace=True)
```

```python
ds.isnull().sum()
```

```python
ds.dropna(subset=['Name','Team','Number','Position','Height','Weight'],inplace=True)
```

```python
ds.isnull().sum()
```

```python
ds['Name'] = ds['Name'].astype('string') 
```

```python
ds['Team'] = ds['Team'].astype('string') 
```

```python
ds['Position'] = ds['Position'].astype('string') 
```

```python
ds['College'] = ds['College'].astype('string') 
```

```python
ds.head()
```

```python
ds.dtypes
```

```python
numeric_cols = ['Age', 'Height', 'Weight', 'Salary']
for col in numeric_cols:
    ds[col] = pd.to_numeric(ds[col], errors='coerce')
```

```python
categorical_var = "Position"
numeric_vars = ["Age", "Weight", "Salary"]
```

```python
grouped_stats = ds.groupby(categorical_var)[numeric_vars].agg(['mean', 'median', 'min', 'max', 'std'])
print(grouped_stats) #Group by categorical variable and get summary statistics
```

```python
position_salary_list = ds.groupby("Position")["Salary"].mean()
print(position_salary_list)
```
