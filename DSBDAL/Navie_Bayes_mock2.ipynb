```python
 #Import required libraries
import pandas as pd
import numpy as np

# For visualization
import matplotlib.pyplot as plt
import seaborn as sns

# For model building
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB

# For evaluation
from sklearn.metrics import confusion_matrix, accuracy_score, classification_report, ConfusionMatrixDisplay
```

```python
# Load the Iris dataset
df = pd.read_csv("Iris.csv")
```

Define Features (X) and Target (y)

Features (X):

    SepalLengthCm
    SepalWidthCm
    PetalLengthCm
    PetalWidthCm

Target (y):

    Species


```python
# Independent variables (features)
X = df.iloc[:, :-1]

# Dependent variable (target)
y = df.iloc[:, -1] #species
```

 Split Dataset (Train-Test Split)(80:20)

```python
# Split dataset
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    random_state=0
)
```

Apply Naïve Bayes Algorithm
We use Gaussian Naïve Bayes because our dataset contains continuous values.

```python
# Create model
model = GaussianNB() #prior probability of class

# Train model
model.fit(X_train, y_train)
```

 Make Predictions

```python
# Predict species
y_pred = model.predict(X_test)
```

```python
print("Predicted values:")
print(y_pred)
```

Evaluate Model

# Create confusion matrix
cm = confusion_matrix(y_test, y_pred)

print("Confusion Matrix:")
print(cm)

```python
cm = confusion_matrix(y_test, y_pred)

print("Confusion Matrix:") 
print(cm)
```

```python
# Visualize Confusion Matrix
disp = ConfusionMatrixDisplay(confusion_matrix = cm)
disp.plot(cmap="Blues")
plt.title("Confusion Matrix")
plt.show()
```

<b>Considering "setosa" as positive class</b>

```python
TP = cm[0][0]
FP = cm[1][0] + cm[2][0]
FN = cm[0][1] + cm[0][2]
TN = cm[1][1] + cm[1][2] + cm[2][1] + cm[2][2]

accuracy = (TP + TN) / (TP + TN + FP + FN)
error_rate = (FP + FN) / (TP + TN + FP + FN)
precision = TP / (TP + FP)
recall = TP / (TP + FN)

print("For Setosa Class:\n")
print("TP:", TP)
print("FP:", FP)
print("TN:", TN)
print("FN:", FN)

print("\nAccuracy:", np.round(accuracy, 2))
print("Error Rate:", np.round(error_rate, 2))
print("Precision:", np.round(precision, 2))
print("Recall:", np.round(recall, 2))
```

<b>Considering "Versiccolor" as positive class</b>

```python
TP = cm[1][1]
FP = cm[0][1] + cm[2][1]
FN = cm[1][0] + cm[1][2]
TN = cm[0][0] + cm[0][2] + cm[2][0] + cm[2][2]

accuracy = (TP + TN) / (TP + TN + FP + FN)
error_rate = (FP + FN) / (TP + TN + FP + FN)
precision = TP / (TP + FP)
recall = TP / (TP + FN)

print("For Versicolor Class:\n")
print("TP:", TP)
print("FP:", FP)
print("TN:", TN)
print("FN:", FN)

print("\nAccuracy:", np.round(accuracy, 2))
print("Error Rate:", np.round(error_rate, 2))
print("Precision:", np.round(precision, 2))
print("Recall:", np.round(recall, 2))
```
