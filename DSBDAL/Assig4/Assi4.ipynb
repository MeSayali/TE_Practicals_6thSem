```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
ds=sns.load_dataset("titanic") #load data from titanic dataset
```

```python
ds.head()
```

```python
ds.info()
```

```python
ds.shape # rows and columns
```

```python
ds.isnull().sum()
```

```python
ds['age'] = ds['age'].fillna(ds['age'].mean()) #handle missing values and replcae with mean
```

```python
ds.isnull().sum()
```

```python
ds['embarked'] = ds['embarked'].fillna(ds['embarked'].mode()[0])
```

```python
ds.isnull().sum()
```

```python
ds.drop(columns=['deck'], inplace=True) #drop column too many missing values
```

```python
ds.isnull().sum()
```

```python
ds.head()
```

```python
sns.countplot(x='sex',data=ds)  #categorical data. 
plt.title("gender Distribution on Titanic")
plt.show()
```

```python
sns.countplot(x='survived',data=ds)
plt.title("Survival Count")
plt.show()
```

```python
sns.barplot(x='sex', y='survived',data=ds) #shows average (mean) of a numerical variable for each category
plt.title("Gender-wise Survival Rate")
plt.show()
```

```python
sns.barplot(x='class',y='survived',data=ds)
plt.title("Class-Wise Survival Rate")
plt.show()
```

```python
sns.histplot(ds['fare'], bins=30)
plt.title("Ticket Fare Distribution")
plt.xlabel("Fare")
plt.ylabel("Count")
plt.show()
```

```python
sns.scatterplot(x='fare', y='survived', data=ds) # to check the 2 numerical variables ka relationship check karna ho
plt.title("Fare vs Survival")
plt.show()
```

```python
sns.countplot(x='pclass', data=ds)
plt.title("Passenger Count by Class")
plt.show()
```

```python
sns.histplot(data=ds, x='age', hue='survived', bins=30)
plt.title("Age Distribution by Survival")
plt.show()
```

```python
sns.boxplot(data=ds,x="survived",y="fare")
```
