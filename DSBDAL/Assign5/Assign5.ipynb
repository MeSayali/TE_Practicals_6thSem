```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
titanic=sns.load_dataset('titanic')
```

```python
titanic_clean=titanic.dropna(subset=['age']) #drop rows with missing age values 
```

```python
sns.set(style="whitegrid")#set plot style
```

```python
plt.figure(figsize=(10, 6)) #create a new figure with width and height
sns.boxplot(
    x='sex',
    y='age',
    hue='survived',
    data=titanic_clean
)
```

```python
# Now add labels, legend, and title
plt.title('Age Distribution by Gender and Survival Status (Titanic Dataset)')
plt.xlabel('Gender')
plt.ylabel('Age')
plt.legend(title='Survived', labels=['No', 'Yes'])

plt.show()
```
