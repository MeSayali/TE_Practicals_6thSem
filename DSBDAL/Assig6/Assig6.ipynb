```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("Iris.csv")
```

```python
df.head()
```

```python
df.describe()
```

```python
df.dtypes #Features and Types
#id=Nominal (Identifier)
#species=Categorical (Nominal)
#remaining are Numeric
```

```python
#Create histograms for each numeric feature
numeric_features = ["SepalLengthCm", "SepalWidthCm",
                    "PetalLengthCm", "PetalWidthCm"]#this all are of numeric datatypes

df[numeric_features].hist(bins=20, figsize=(10, 8), edgecolor='black')# first select numeric column then hist is function to craete histogram bins =20 as interval
plt.suptitle("Histogram of Iris Features")
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(data=df[numeric_features])
plt.title("Box Plot of Iris Features")
plt.show()
#x axis contain names 
#y axis shows the value of features for ex:SepalLegth range from 4-8
```

```python
#Box plots of features with respect to species like iris-setosa etc.
plt.figure(figsize=(10, 6))

# Box plot for SepalLengthCm

sns.boxplot(x="Species", y="SepalLengthCm", data=df)
plt.title("SepalLengthCm")
```

```python
#Box plots of features with respect to species like iris-setosa etc.
plt.figure(figsize=(10, 8))

# Box plot for SepalLengthCm

sns.boxplot(x="Species", y="SepalWidthCm", data=df)
plt.title("SepalWidthCm")
```

```python
#Box plots of features with respect to species like iris-setosa etc.
plt.figure(figsize=(10, 6))

# Box plot for SepalLengthCm

sns.boxplot(x="Species", y="PetalLengthCm", data=df)
plt.title("PetalLengthCm")
```
