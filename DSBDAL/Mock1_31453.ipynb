Now we will Remove this null values by replcaing it with the Mean of that column


```python
dataset.fillna({'Car':dataset['Car'].mean()},inplace=True)
```

```python
dataset.fillna({'YearBuilt':dataset['YearBuilt'].mean()},inplace=True)
```

```python
dataset.fillna({'BuildingArea':dataset['BuildingArea'].mean()},inplace=True)
```

```python
scaler = MinMaxScaler() #normalizes numerical columns
num_cols = ['Price', 'Distance', 'Landsize', 'BuildingArea']# Select numeric columns to scale
dataset[num_cols] = scaler.fit_transform(dataset[num_cols]) # Apply Min-Max Scaling
```
